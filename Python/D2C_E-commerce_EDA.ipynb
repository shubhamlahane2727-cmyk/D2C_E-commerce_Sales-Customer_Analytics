# 📌 Project Overview

The Direct-to-Consumer (D2C) e-commerce industry generates massive amounts of customer, product, order, review, and website interaction data every day. Converting this raw data into meaningful business insights helps organizations improve customer experience, optimize product performance, increase revenue, and make data-driven decisions.

This project performs a complete Exploratory Data Analysis (EDA) on a D2C E-commerce database using **Python** by directly connecting to a **MySQL** database. Instead of importing CSV files, all datasets are retrieved using SQL queries, simulating a real-world business analytics workflow where analysts work with production databases.

The project analyzes the complete customer journey—from customer registration and product browsing to purchasing behavior, customer reviews, and website engagement—to uncover actionable business insights.

---

# 🎯 Project Objectives

The objectives of this project are to:

- Analyze overall sales performance and revenue trends.
- Identify top-selling products, brands, and categories.
- Understand customer purchasing behavior and spending patterns.
- Analyze monthly and yearly business growth.
- Evaluate customer reviews and product ratings.
- Study customer website interactions (views, cart, purchases).
- Discover business opportunities through data-driven insights.
- Create professional visualizations for business reporting.

---

# 🗂 Database Information

The analysis is performed on a MySQL database consisting of six relational tables.

| Table | Description |
|--------|-------------|
| Users | Customer information and signup details |
| Products | Product details including category, brand, and price |
| Orders | Customer order information |
| Order_Items | Individual products purchased in each order |
| Reviews | Customer ratings and reviews |
| Events | Customer website interactions (View, Cart, Purchase) |

## Business Analysis

## 1. Sales Analysis

````python
total_revenue = orders['total_amount'].sum()

print(f"Total Revenue: ₹{total_revenue:,.2f}")
````

````python
total_orders = orders['order_id'].nunique()

print("Total Orders:", total_orders)
````

````python
aov = orders['total_amount'].mean()

print(f"Average Order Value: ₹{aov:.2f}")
````

````python
monthly_revenue = (
    orders
    .groupby(orders['order_date'].dt.to_period('M'))['total_amount']
    .sum()
)

monthly_revenue.index = monthly_revenue.index.astype(str)

plt.figure(figsize=(12,5))

plt.plot(monthly_revenue.index,
         monthly_revenue.values,
         marker='o')

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
````

````python
monthly_orders = (
    orders
    .groupby(orders['order_date'].dt.to_period('M'))
    .size()
)

monthly_orders.index = monthly_orders.index.astype(str)

plt.figure(figsize=(12,5))

plt.plot(monthly_orders.index,
         monthly_orders.values,
         marker='o')

plt.title("Monthly Orders Trend")
plt.xlabel("Month")
plt.ylabel("Orders")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
````

````python
order_status = orders['order_status'].value_counts()

plt.figure(figsize=(8,5))

order_status.plot(kind='bar')

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Orders")

plt.show()
````

````python
status_revenue = (
    orders
    .groupby('order_status')['total_amount']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

status_revenue.plot(kind='bar')

plt.title("Revenue by Order Status")
plt.xlabel("Order Status")
plt.ylabel("Revenue")

plt.show()
````

````python
city_revenue = (
    orders
    .merge(users,on='user_id')
    .groupby('city')['total_amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))

city_revenue.plot(kind='bar')

plt.title("Top 10 Cities by Revenue")
plt.xlabel("City")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
````

````python
plt.figure(figsize=(8,5))

plt.hist(orders['total_amount'], bins=30)

plt.title("Revenue Distribution")
plt.xlabel("Order Amount")
plt.ylabel("Frequency")

plt.show()
````

## 4. Review Analysis

````python
total_reviews = reviews['review_id'].nunique()

print("Total Reviews:", total_reviews)
````

````python
avg_rating = reviews['rating'].mean()

print(f"Average Rating: {avg_rating:.2f}")
````

````python
plt.figure(figsize=(8,5))

reviews['rating'].value_counts().sort_index().plot(kind='bar')

plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Reviews")

plt.show()
````

````python
monthly_reviews = (
    reviews
    .groupby(reviews['review_date'].dt.to_period('M'))
    .size()
)

monthly_reviews.index = monthly_reviews.index.astype(str)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_reviews.index,
    monthly_reviews.values,
    marker='o'
)

plt.title("Monthly Review Trend")
plt.xlabel("Month")
plt.ylabel("Number of Reviews")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
````

````python
top_reviewed = (
    reviews
    .groupby('product_id')['review_id']
    .count()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

top_reviewed = top_reviewed.merge(
    products[['product_id','product_name']],
    on='product_id'
)

plt.figure(figsize=(12,5))

plt.bar(
    top_reviewed['product_name'],
    top_reviewed['review_id']
)

plt.title("Top 10 Most Reviewed Products")
plt.xlabel("Product")
plt.ylabel("Number of Reviews")

plt.xticks(rotation=45, ha='right')

plt.show()
````

````python
highest_rated = (
    reviews
    .groupby('product_id')
    .agg(
        Average_Rating=('rating','mean'),
        Total_Reviews=('review_id','count')
    )
)

highest_rated = highest_rated[
    highest_rated['Total_Reviews'] >= 5
]

highest_rated = highest_rated.sort_values(
    'Average_Rating',
    ascending=False
).head(10)

highest_rated = highest_rated.merge(
    products[['product_id','product_name']],
    on='product_id'
)

plt.figure(figsize=(12,5))

plt.bar(
    highest_rated['product_name'],
    highest_rated['Average_Rating']
)

plt.title("Top Rated Products")
plt.xlabel("Product")
plt.ylabel("Average Rating")

plt.xticks(rotation=45, ha='right')

plt.show()
````

````python
rating_category = (
    reviews
    .merge(products, on='product_id')
    .groupby('category')['rating_x']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

rating_category.plot(kind='bar')

plt.title("Average Rating by Category")
plt.xlabel("Category")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)

plt.show()
````

````python
rating_brand = (
    reviews
    .merge(products, on='product_id')
    .groupby('brand')['rating_x']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))

rating_brand.plot(kind='bar')

plt.title("Top 10 Brands by Average Rating")
plt.xlabel("Brand")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)

plt.show()
````

````python
reviews['Sentiment'] = reviews['rating'].apply(
    lambda x: 'Positive' if x >= 4
    else ('Neutral' if x == 3 else 'Negative')
)

sentiment = reviews['Sentiment'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    sentiment,
    labels=sentiment.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Review Sentiment Distribution")

plt.show()
````
